```python
# business questions
# 1. department wise avg salary
df.groupby("department")["salary"].mean()
```

```python
# 2. city wise avg salary
df.groupby("city")["salary"].mean()
```

```python
# 3.How many employees are there in each department
df["department"].value_counts()
```

```python
# 4.Top 5 highest-paid employees
df.sort_values("salary",ascending=False).head(5)
```

```python
# 5.How many employees are there in each city
df["city"].value_counts()
```
